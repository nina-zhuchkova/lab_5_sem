```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from scipy.stats import linregress
import matplotlib
import scipy.optimize as opt
matplotlib.rcParams.update({'font.size': 10})
matplotlib.rcParams['lines.linewidth'] = 2
plt.rcParams['axes.formatter.limits'] = (-2, 2)


light_green = (0.4, 0.8, 0.2)
dark_purple = (0.4, 0.3, 0.7)
raspberry_pink = (1, 0, 0.5)
```

```python
path = 'спектры излучения/i55.esp'
nm = 1e-9
df = pd.read_csv(path, 
                 skiprows=2,           # строки с параметрами
                 sep=r'\s+',           # разделитель — пробелы
                 header=None,          # нет заголовка
                 names=['lambda, nm', 'counts'])  # даем имена колонкам

df_cut = df[df['counts'] > 1e3]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

fig.suptitle(path)

ax1.plot(df['lambda, nm']*nm, df['counts'])
ax1.set_xlabel(r'$\lambda$, m')
ax1.set_ylabel('counts per second')
ax1.ticklabel_format(style='sci')
ax1.set_title('Полный спектр')
ax1.grid(c = (0.85, 0.85, 0.85), ls = '--', lw = 2)

ax2.plot(df_cut['lambda, nm']*nm, df_cut['counts'])
ax2.set_xlabel(r'$\lambda$, m')
ax2.set_ylabel('counts per second')
ax2.ticklabel_format(style='sci')
ax2.set_title('Обрезанный спектр')
ax2.grid(c = (0.85, 0.85, 0.85), ls = '--', lw = 2)
```
